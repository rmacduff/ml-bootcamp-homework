```python
# Q7. Sum of weights
```

```python
cars_asian = df[df['origin'] == 'Asia']
```

```python
cars_asian = cars_asian[['vehicle_weight', 'model_year']]
```

```python
X = cars_asian.head(n=7).values
```

```python
X
```

```python
X.T
```

```python
XTX = X.T.dot(X)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
w = XTX_inv.dot(X.T) * y
```

```python
w.sum()
```
